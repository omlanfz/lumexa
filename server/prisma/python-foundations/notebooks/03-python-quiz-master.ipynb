## Part 1 · The questions

Each question is a **tuple** of three things: the text, a **list** of four options and the number (1–4) of the correct option. All the questions live in one list, so adding a question means adding one line.

```python
QUESTIONS = [
    ("Which keyword defines a function in Python?", ["func", "def", "define", "function"], 2),
    ("What does print(2 ** 3) display?", ["6", "8", "9", "23"], 2),
    ("Which planet is known as the Red Planet?", ["Venus", "Jupiter", "Mars", "Mercury"], 3),
    ("What data type is the value 3.14?", ["int", "str", "float", "bool"], 3),
    ("Which loop repeats while a condition is true?", ["for", "while", "if", "def"], 2),
    ("What is the result of len('Rocket')?", ["5", "6", "7", "Rocket"], 2),
    ("Which symbol starts a comment in Python?", ["//", "#", "--", "/*"], 2),
    ("What does a return statement do?", ["Prints a value", "Ends the program", "Sends a value back from a function", "Repeats a loop"], 3),
    ("In 1961, who became the first human in space?", ["Neil Armstrong", "Yuri Gagarin", "Buzz Aldrin", "Alan Shepard"], 2),
    ("Which collection stores key-value pairs?", ["list", "tuple", "set", "dictionary"], 4),
]

print("Number of questions:", len(QUESTIONS))
text, options, correct = QUESTIONS[0]          # unpack one tuple
print("First question:", text)
print("Options:", options, "| correct option:", correct)
```

## Part 2 · A function that asks one question

`ask_question()` prints the question and the numbered options, then keeps asking **until the answer is valid**: a whole number from 1 to 4. It **returns** `True` or `False` so that the rest of the program can use the result.

In demo mode the notebook plays a scripted game. Look for the player's deliberate typing mistake (`x`): the program does not mark it wrong, it just asks again.

```python
demo_answers = iter(["2", "2", "x", "3", "3", "2", "2", "2", "3", "1", "4"])   # one wrong answer + one typing mistake

def ask_question(number, text, options, correct):
    """Ask one question. Return True if the player is right, otherwise False."""
    print(f"\nQuestion {number}: {text}")
    for index, option in enumerate(options, start=1):
        print(f"  {index}. {option}")

    while True:                                    # keep asking until the answer is valid
        answer = ask("Your answer (1-4): ", next(demo_answers, "1")).strip()
        if answer.isdigit() and 1 <= int(answer) <= len(options):
            return int(answer) == correct
        print("  ⚠️  Please type a number from 1 to 4.")

print(ask_question(1, *QUESTIONS[0]))              # the * unpacks the tuple into the parameters
```

## Part 3 · Feedback and ranks

Two more small functions, each with **one job**:

- `show_feedback(is_correct, right_text)` prints a different message for right and wrong answers.
- `rank_for(percent)` **returns** a rank name. A function that only *returns* a value (and prints nothing) is easy to test.

```python
def show_feedback(is_correct, right_text):
    if is_correct:
        print("  ✅ Correct!")
    else:
        print(f"  ❌ Not quite. The right answer was: {right_text}")

def rank_for(percent, pass_mark=50):
    """Return a rank name for a percentage."""
    if percent >= 90:
        return "Commander 🚀"
    elif percent >= 70:
        return "Pilot ✈️"
    elif percent >= pass_mark:
        return "Cadet 🎓"
    else:
        return "Rookie 🌱 (keep practising!)"

# Test the function on its own before using it
for test_percent in [100, 75, 50, 20]:
    print(f"{test_percent}% -> {rank_for(test_percent)}")
```

## Part 4 · The whole quiz

`run_quiz()` loops over the questions, calls the small functions, adds up the score and shows the result. A `main()` function starts everything. This is how real programs are organised: **many small functions and one function that calls them in order**.

```python
def show_results(score, total):
    percent = score / total * 100
    print("\n" + "=" * 34)
    print(f"  You scored {score} out of {total}  ({percent:.0f}%)")
    print(f"  Your rank: {rank_for(percent)}")
    print("=" * 34)

def run_quiz(questions):
    """Ask every question and return the score."""
    score = 0
    for number, (text, options, correct) in enumerate(questions, start=1):
        is_correct = ask_question(number, text, options, correct)
        show_feedback(is_correct, options[correct - 1])
        if is_correct:
            score += 1
    return score

def main():
    print("🚀 WELCOME TO THE PYTHON QUIZ MASTER 🚀")
    score = run_quiz(QUESTIONS)
    show_results(score, len(QUESTIONS))

demo_answers = iter(["2", "2", "x", "3", "3", "2", "2", "2", "3", "1", "4"])   # reset the scripted player
main()
```

3. Check your questions: each one needs exactly **four** options, and the number of the right one must be between 1 and 4. The checker below counts any problems for you.

```python
def check_questions(questions):
    """Return a list of problems found in the question list."""
    problems = []
    for number, (text, options, correct) in enumerate(questions, start=1):
        if len(options) != 4:
            problems.append(f"Question {number} does not have 4 options")
        if not 1 <= correct <= len(options):
            problems.append(f"Question {number} has an impossible correct answer: {correct}")
    return problems

problems = check_questions(QUESTIONS)
print("✅ All questions look good!" if not problems else problems)
```

## Part 5 · Level up: shuffle and choose a length

A quiz that always asks the same questions in the same order gets boring. `random.shuffle()` mixes up a list. We will also let the player choose how many questions to answer, using a **default parameter**.

```python
def run_short_quiz(questions, how_many=5):
    """Shuffle the questions and ask only how_many of them."""
    mixed = questions[:]                      # a copy, so the original order is kept
    random.shuffle(mixed)
    return run_quiz(mixed[:how_many])

if not INTERACTIVE:
    random.seed(5)
demo_answers = iter(["2", "1", "3", "2", "4"])
score = run_short_quiz(QUESTIONS, how_many=5)
show_results(score, 5)
```
