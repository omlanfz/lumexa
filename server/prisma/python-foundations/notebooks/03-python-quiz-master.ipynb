A customizable quiz application with questions, answer checking and scoring.

* **Part 1 · Lesson 6:** string methods that clean and check the answers.
* **Part 2 · Lesson 7:** lists, tuples, sets and dictionaries for the questions, the score and the topics.

```python
# ======================================================================
#  PYTHON QUIZ MASTER
#  Lumexa Python Foundations · Course 1: Python Programming Essentials
#  Part 1 · Lesson 6: strings (clean text, compare answers, hints)
#  Part 2 · Lesson 7: collections (lists, tuples, sets and dictionaries)
# ======================================================================


# ----------------------------------------------------------------------
# PART 1 · Lesson 6: working with strings
# ----------------------------------------------------------------------
print("=" * 40)
print("PYTHON QUIZ MASTER - PART 1")
print("=" * 40)

# strip() removes spaces at both ends, title() capitalises every word.
name = input("What is your name? ").strip().title()
print("Welcome, " + name + "! Your name has", len(name), "letters.")
print("Get ready, " + name.upper() + "!")
print("Your name backwards is", name[::-1])

warmup_points = 0

# Question 1: compare CLEAN text. lower() ignores capital letters, and `in` finds a word inside a longer answer.
print()
print("Question 1: Which function shows text on the screen?")
answer = input("Your answer: ").strip().lower()
correct_answer = "print"
if answer == correct_answer:
    print("Correct!")
    warmup_points += 1
elif correct_answer in answer:
    print("Almost! You typed", answer, "but the answer is just", correct_answer.upper() + ".")
else:
    print("Not quite. The answer is", correct_answer.upper() + ".")

# Question 2: a hint built from string tools (indexing, repeating, slicing, count).
secret_word = "variable"
hint = secret_word[0] + "_" * (len(secret_word) - 1)
print()
print("Question 2: I store a value under a name. What am I?")
print("Hint:", hint, "(" + str(len(secret_word)), "letters)")
answer = input("Your answer: ").strip().lower().replace("-", "")
if answer == secret_word:
    print("Correct!")
    warmup_points += 1
else:
    print("Not quite. The answer is", secret_word.title() + ".")
    print("It has", secret_word.count("a"), "letter a's, the letter b first appears at index", secret_word.find("b"), "and starts with", secret_word[:3] + ".")

print()
print(f"Warm-up score: {warmup_points} of 2")


# ----------------------------------------------------------------------
# PART 2 · Lesson 7: working with collections
# ----------------------------------------------------------------------
print()
print("=" * 40)
print("PYTHON QUIZ MASTER - PART 2")
print("=" * 40)

# A list of tuples: every question is a tuple (question, answer, hint, topic) that never changes.
questions = [
    ("What does len([4, 8, 15]) return?", "3", "Count the items.", "lists"),
    ("Which collection keeps only unique values?", "set", "Duplicates disappear.", "sets"),
    ("Which collection stores key-value pairs?", "dictionary", "Think of a real dictionary.", "dictionaries"),
    ("Which list method adds an item to the end?", "append", "It starts with the letter a.", "lists"),
    ("Which collection cannot be changed once it is made?", "tuple", "It starts with the letter t.", "tuples"),
]
ranks = (("Commander", 90), ("Pilot", 70), ("Cadet", 0))    # a tuple of tuples, best rank first

score = 0
max_score = len(questions) * 2
wrong_answers = []        # a list that grows as the player makes mistakes
results = {}              # a dictionary: question -> True (solved) or False
topic_points = {}         # a dictionary: topic -> points earned
topics = set()            # a set: every topic once, however many questions use it

number = 1
for question, correct, hint, topic in questions:
    print()
    print(f"Question {number} of {len(questions)} ({topic}): {question}")
    topics.add(topic)
    attempts = 0
    solved = False
    while attempts < 2 and not solved:
        reply = input("Your answer: ").strip().lower()
        attempts += 1
        if reply == correct:
            solved = True
        elif attempts == 1:
            print("Not quite. Hint:", hint)

    results[question] = solved
    if solved and attempts == 1:
        points = 2
    elif solved:
        points = 1
    else:
        points = 0
        wrong_answers.append(correct)
    score += points
    topic_points[topic] = topic_points.get(topic, 0) + points
    print("Points for this question:", points)
    number += 1

# The results screen: the score, a percentage and a rank from the tuple of ranks.
percentage = score / max_score * 100
rank = "Cadet"
for rank_name, minimum in ranks:
    if percentage >= minimum:
        rank = rank_name
        break

solved_count = 0
for was_solved in results.values():
    if was_solved:
        solved_count += 1

print()
print("=" * 40)
print(f"RESULTS FOR {name.upper()}")
print("=" * 40)
print(f"Questions solved: {solved_count} of {len(questions)}")
print(f"Score: {score} of {max_score} ({percentage:.0f}%)")
print("Rank:", rank)
print("Topics covered:", ", ".join(sorted(topics)))
for topic in sorted(topic_points):
    print("  " + topic + ":", topic_points[topic], "points")
if len(wrong_answers) > 0:
    print("Review these answers:", ", ".join(wrong_answers))
else:
    print("Perfect round: nothing to review!")
```
