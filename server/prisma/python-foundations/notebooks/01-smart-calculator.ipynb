## Part 2 · All seven operators at once

Python has seven arithmetic operators. We use an **f-string** for every line, and `:.2f` to show two decimal places.

| Operator | Meaning |
|---|---|
| `+` `-` `*` | add, subtract, multiply |
| `/` | divide (always a decimal number) |
| `//` | floor division: the whole-number part |
| `%` | modulo: the remainder |
| `**` | power |

```python
print(f"{first} + {second} = {first + second:.2f}")
print(f"{first} - {second} = {first - second:.2f}")
print(f"{first} * {second} = {first * second:.2f}")
print(f"{first} / {second} = {first / second:.2f}")
print(f"{first} // {second} = {first // second:.2f}   (whole part of the division)")
print(f"{first} % {second} = {first % second:.2f}   (what is left over)")
print(f"{first} ** {second} = {first ** second:.2f}   (first to the power of second)")
```

## Part 3 · The "smart" extras

A calculator becomes *smart* when it solves everyday problems. Here are four, each using operators and a variable you can change.

```python
# A discount
price = float(ask("Price of an item: ", 80))
discount_percent = float(ask("Discount in percent: ", 15))
final_price = price - price * discount_percent / 100
print(f"Final price: {final_price:.2f}  (you save {price - final_price:.2f})")

# An average of three test scores
score1 = float(ask("Score 1: ", 78))
score2 = float(ask("Score 2: ", 91))
score3 = float(ask("Score 3: ", 84))
average = (score1 + score2 + score3) / 3          # brackets first!
print(f"Average score: {average:.1f}")

# Using += to add a bonus point
average += 1
print(f"Average with a bonus point: {average:.1f}")

# A temperature converter
celsius = float(ask("Temperature in Celsius: ", 36.6))
print(f"{celsius}°C = {celsius * 9 / 5 + 32:.1f}°F")

# A comparison gives True or False
print("Is the first number bigger than the second?", first > second)
```

## Part 4 · Let the user choose (after Lesson 4: `if`, `elif`, `else`)

Now the calculator becomes truly smart: the user **chooses** the operation. We use an `if` / `elif` / `else` chain, and we protect against dividing by zero.

> You can run this part now and read how it works. You will write decisions like these yourself in Lesson 4.

```python
print("Operations: +  -  *  /  //  %  **")
operation = ask("Choose an operation: ", "/").strip()
a = float(ask("First number: ", 22))
b = float(ask("Second number: ", 0))

if operation == "+":
    result = a + b
elif operation == "-":
    result = a - b
elif operation == "*":
    result = a * b
elif operation in ("/", "//", "%") and b == 0:
    result = None
    print("⚠️  You cannot divide by zero. Try a different second number.")
elif operation == "/":
    result = a / b
elif operation == "//":
    result = a // b
elif operation == "%":
    result = a % b
elif operation == "**":
    result = a ** b
else:
    result = None
    print("⚠️  I do not know the operation", repr(operation))

if result is not None:
    print(f"{a} {operation} {b} = {result:.2f}")
```

## Part 5 · A full calculator session (after Lesson 5: loops)

A real calculator keeps running until you quit, and remembers what you did. A `while True` loop repeats the menu, `break` leaves it, and a **list** (Lesson 7) keeps a history.

In demo mode the notebook plays a scripted session, including two mistakes, so you can see how the program copes.

```python
import operator

# A dictionary of functions: each operator symbol points to the tool that does the maths.
# (You will meet functions in Lesson 8 and dictionaries in Lesson 7.)
OPERATIONS = {"+": operator.add, "-": operator.sub, "*": operator.mul, "/": operator.truediv,
              "//": operator.floordiv, "%": operator.mod, "**": operator.pow}

history = []
demo_session = iter([("+", 12, 30), ("/", 9, 0), ("pizza", 1, 1), ("**", 2, 10), ("q", 0, 0)])

while True:
    demo = next(demo_session, ("q", 0, 0))
    op = ask("Operation (or q to quit): ", demo[0]).strip()
    if op == "q":
        print("Goodbye, Commander! 👋")
        break
    if op not in OPERATIONS:
        print("⚠️  Unknown operation. Try again.\n")
        continue
    x = float(ask("First number: ", demo[1]))
    y = float(ask("Second number: ", demo[2]))
    if op in ("/", "//", "%") and y == 0:
        print("⚠️  Cannot divide by zero.\n")
        continue
    answer = OPERATIONS[op](x, y)
    line = f"{x} {op} {y} = {answer:.2f}"
    history.append(line)
    print(line, "\n")

print("History of this session:")
for number, line in enumerate(history, start=1):
    print(f"  {number}. {line}")
```
