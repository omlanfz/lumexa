# Smart Calculator

An interactive calculator that supports different mathematical operations.

* **Part 1 · Lesson 2:** variables, data types and converting values.
* **Part 2 · Lesson 3:** `input()`, f-strings and operators.

```python
# ======================================================================
#  SMART CALCULATOR
#  Lumexa Python Foundations · Course 1: Python Programming Essentials
#  Part 1 · Lesson 2: variables, data types and values
#  Part 2 · Lesson 3: user input, output and operators
# ======================================================================


# ----------------------------------------------------------------------
# PART 1 · Lesson 2: variables, data types and values
# ----------------------------------------------------------------------
# A variable is a name that points to a value. Python has four basic kinds of value.

print("========================================")
print("   SMART CALCULATOR - PART 1")
print("========================================")

first_number = 12        # int   (a whole number)
second_number = 4.5      # float (a decimal number)
operation = "addition"   # str   (text)
show_types = True        # bool  (True or False)

print("first_number =", first_number, "| type:", type(first_number))
print("second_number =", second_number, "| type:", type(second_number))
print("operation =", operation, "| type:", type(operation))
print("show_types =", show_types, "| type:", type(show_types))

# Calculations use the values stored in the variables.
total = first_number + second_number
print("The", operation, "of", first_number, "and", second_number, "is", total)
print("A whole number plus a decimal gives a:", type(total))

# Types matter: "12" (text) is not the same as 12 (a number).
text_number = "12"
print('"12" + "12" joins the text:', text_number + text_number)
print("12 + 12 adds the numbers:", first_number + first_number)

# Convert between types with int(), float() and str().
price_text = "250"
price = int(price_text)
print("The price plus 50:", price + 50)
print("The price as text again: " + str(price))
print("Turning 4.5 into a whole number drops the decimals:", int(second_number))

# A variable can be given a new value (reassigned) at any time.
running_total = 0
running_total = running_total + first_number
running_total = running_total + second_number
print("Running total:", running_total)

# The four basic operators, used on variables.
print("add:", first_number + second_number)
print("subtract:", first_number - second_number)
print("multiply:", first_number * second_number)
print("divide:", first_number / second_number)


# ----------------------------------------------------------------------
# PART 2 · Lesson 3: user input, output and operators
# ----------------------------------------------------------------------
print()
print("========================================")
print("   SMART CALCULATOR - PART 2")
print("========================================")

# input() always gives back text, so we convert it with float() before calculating.
# Type a second number that is not 0: dividing by zero stops the program.
# (In Lesson 4 you will learn how to prevent that with an if statement.)
a = float(input("First number: "))
b = float(input("Second number (not 0): "))

# All seven arithmetic operators, shown with f-strings and two decimal places.
print()
print(f"{a} + {b} = {a + b:.2f}")
print(f"{a} - {b} = {a - b:.2f}")
print(f"{a} * {b} = {a * b:.2f}")
print(f"{a} / {b} = {a / b:.2f}")
print(f"{a} // {b} = {a // b:.2f}   (the whole part of the division)")
print(f"{a} % {b} = {a % b:.2f}   (what is left over)")
print(f"{a} ** {b} = {a ** b:.2f}   (a to the power of b)")

# sep and end change how print() joins its items and how it finishes the line.
print("a", "b", "c", sep="-")
print("Loading", end="...")
print("done!")

# Smart extra 1: a discount.
price = float(input("Price of an item: "))
percent = float(input("Discount in percent: "))
final_price = price - price * percent / 100
print(f"Final price: {final_price:.2f} (you save {price - final_price:.2f})")

# Smart extra 2: the average of three test scores (brackets decide what is calculated first).
score1 = float(input("Test score 1: "))
score2 = float(input("Test score 2: "))
score3 = float(input("Test score 3: "))
average = (score1 + score2 + score3) / 3
print(f"Average: {average:.1f}")
average += 1                      # augmented assignment: add a bonus point
print(f"With a bonus point: {average:.1f}")

# Smart extra 3: a temperature converter.
celsius = float(input("Temperature in Celsius: "))
print(f"{celsius}°C = {celsius * 9 / 5 + 32:.1f}°F")

# Comparison and logical operators give True or False. Precedence: * before +.
print("Is the first number bigger than the second?", a > b)
print("Is the first number between 0 and 100?", a >= 0 and a <= 100)
print("2 + 3 * 4 =", 2 + 3 * 4, "but (2 + 3) * 4 =", (2 + 3) * 4)

print()
print("Thank you for using the Smart Calculator!")
```
